```python
import pandas as pd
import numpy as np
```

```python
#Load the dataset to a pandas dataframe named "df"
filepath = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/laptop_pricing_dataset_base.csv"
df = pd.read_csv(filepath, header=None)
```

```python
#print the firstg 5 entries of the dataset to confirm loading
print(df.head())
```

```python
# create headers list
headers = ["Manufacturer", "Category", "Screen", "GPU", "OS", "CPU_core", "Screen_Size_inch", "CPU_frequency", "RAM_GB", "Storage_GB_SSD", "Weight_kg", "Price"]
df.columns = headers
print(df.head(10))
```

```python
#Replace the "?" entries in the dataset with NaN value
df.replace('?',np.nan, inplace = True)
```

```python
#Make a note of the data types of the different columns of the dataset
print(df.dtypes)
```

```python
#Print the statistical description of the dataset, including that of 'object' data types
print(df.describe(include='all'))
```

```python
#Print the summary info of dataset
print(df.info())
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df[['Screen_Size_inch']] = np.round(df[['Screen_Size_inch']],2)
df.head()
```

```python
#Evaluate the dataset for missing data
missing_data = df.isnull()
print(missing_data.head())
for column in missing_data.columns.values.tolist():
    print(column)
    print (missing_data[column].value_counts())
    print("")  
```

```python
#Replace missing data with Mean value
avg_weight=df['Weight_kg'].astype('float').mean(axis=0)
df["Weight_kg"].replace(np.nan, avg_weight, inplace=True)

# astype() function converts the values to the desired data type
# axis=0 indicates that the mean value is to calculated across all column elements in a row.
```

```python
# replacing missing data with mode
common_screen_size = df['Screen_Size_inch'].value_counts().idxmax()
df["Screen_Size_inch"].replace(np.nan, common_screen_size, inplace=True)
```

```python
df[["Weight_kg","Screen_Size_inch"]] = df[["Weight_kg","Screen_Size_inch"]].astype("float")
```

```python
# Data standardization: convert weight from kg to pounds
df["Weight_kg"] = df["Weight_kg"]*2.205
df.rename(columns={'Weight_kg':'Weight_pounds'}, inplace=True)
```

```python
df['CPU_frequency'] = df['CPU_frequency']/df['CPU_frequency'].max()
```

```python
#Create bins for high, medium, and low pricing
bins = np.linspace(min(df["Price"]), max(df["Price"]), 4)
group_names = ['Low', 'Medium', 'High']
df['Price-binned'] = pd.cut(df['Price'], bins, labels=group_names, include_lowest=True )
```

```python
#Plot the bar graph of these bins
plt.bar(group_names, df["Price-binned"].value_counts())
plt.xlabel("Price")
plt.ylabel("count")
plt.title("Price bins")
```

```python
#Indicator Variable: Screen
dummy_variable_1 = pd.get_dummies(df["Screen"])
dummy_variable_1.rename(columns={'IPS Panel':'Screen-IPS_panel', 'Full HD':'Screen-Full_HD'}, inplace=True)
df = pd.concat([df, dummy_variable_1], axis=1)

# drop original column "Screen" from "df"
df.drop("Screen", axis = 1, inplace=True)
```
